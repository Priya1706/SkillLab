```python
#bubble sort
a=[6,5,3,7,9,1]
while True:
    b = False
    for i in range(len(a)-1):
        if a[i] > a[i+1]:
            a[i], a[i+1] = a[i+1], a[i]
            b = True
    if not b:
        break
print("Sorted list:", a)
```

```python
marks = []
n=str(input("Enter your name: "))
b=str(input("Enter your branch: "))
for i in range(5):
    print(f"Subject {i+1}: ")
    g = float(input("Enter grade: "))
    marks.append(g)
print("====STUDENT REPORT====")
print("Name:", n)
print("Branch:", b)
for i in range(5):
    print(f"Subject {i+1}:",marks[i])
print("total:", sum(marks))
avg = sum(marks)/len(marks)
print("average:",avg)
if avg >= 40:
    print("pass")
else:
    print("fail")
```
